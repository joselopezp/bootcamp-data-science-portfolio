## KPI definitions used in this project

- **Buyer:** customer with at least one transaction (392 of 500 registered; 108 never purchased).
- **Retargeting segment**, by days since the last purchase, measured from the last transaction in the data (January 2025): Active up to 90 days, At Risk 90–180 days, Dormant 180+ days.
- **Dormant rate** is a proxy for churn: in a non-contractual business, a customer who stops buying is never observed leaving. Customer rates use buyers as the denominator and state it.
- **NPS:** answered by buyers only. In this dataset the survey is simulated (see the Post-Evaluation section of the README).
